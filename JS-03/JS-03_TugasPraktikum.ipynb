# **NO 1 - Pemisahan Variabel**

```python
# Menghapus kolom yang tidak digunakan
df = df.drop(columns=['id', 'Unnamed: 32', ])

df.head()
```

# **No. 2 - Encoding Kolom diagnosis**

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])
```

```python
df.head(100)
```

# **NO 3 - Standardisasi kolom numerik.**

```python
from sklearn.preprocessing import StandardScaler

std = StandardScaler()

kolom_numerik = df.drop(columns=['diagnosis']).columns
df[kolom_numerik] = std.fit_transform(df[kolom_numerik])

df.head(100)
```

# **NO 4 - Stratified Split Data**

```python
# Split data
from sklearn.model_selection import train_test_split

df_train, df_test = train_test_split(df, test_size=0.2, random_state=0, stratify=df['diagnosis'])

# Cek masing-masing ukuran data

print(f'Jumlah label data asli:\n{df.diagnosis.value_counts()}')
print(f'Jumlah label data train:\n{df_train.diagnosis.value_counts()}')
print(f'Jumlah label data test:\n{df_test.diagnosis.value_counts()}')
```
