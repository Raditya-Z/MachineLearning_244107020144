3. Modeling


3.1 Penentuan Jumlah Klaster dengan Elbow Method


### Fungsi K-Means Manhattan

Implementasi menggunakan **Manhattan Distance (L1)**. Mengacu pada notebook contoh, pusat cluster diperbarui menggunakan **median setiap fitur (K-Medians Principle)** karena median sesuai untuk meminimalkan jumlah absolute error pada Manhattan Distance. Struktur eksperimen dibuat sama dengan K-Means Euclidean dan K-Means Cosine agar hasil dapat dibandingkan secara konsisten.


```python
class KMeansManhattan:
    def __init__(
        self,
        n_clusters=3,
        max_iter=300,
        tol=1e-4,
        random_state=42
    ):
        self.n_clusters = n_clusters
        self.max_iter = max_iter
        self.tol = tol
        self.random_state = random_state
        self.cluster_centers_ = None
        self.labels_ = None
        self.inertia_ = 0.0
        self.n_iter_ = 0

    def _compute_distance(
        self,
        X,
        centroids
    ):
        return pairwise_distances(
            X,
            centroids,
            metric='manhattan'
        )

    def _init_centroids(self, X):
        rng = np.random.RandomState(
            self.random_state
        )

        n_samples = X.shape[0]

        first_idx = rng.randint(
            n_samples
        )

        centroids = [
            X[first_idx]
        ]

        for _ in range(
            1,
            self.n_clusters
        ):
            current_centroids = np.array(
                centroids
            )

            distances = self._compute_distance(
                X,
                current_centroids
            )

            min_distances = np.min(
                distances,
                axis=1
            )

            squared_distances = (
                min_distances ** 2
            )

            total = np.sum(
                squared_distances
            )

            if total == 0:
                probabilities = (
                    np.ones(n_samples)
                    / n_samples
                )
            else:
                probabilities = (
                    squared_distances
                    / total
                )

            next_idx = rng.choice(
                n_samples,
                p=probabilities
            )

            centroids.append(
                X[next_idx]
            )

        return np.array(
            centroids
        )

    def fit(self, X):
        X_array = np.asarray(
            X,
            dtype=float
        )

        centroids = self._init_centroids(
            X_array
        )

        rng = np.random.RandomState(
            self.random_state
        )

        for iteration in range(
            self.max_iter
        ):
            distances = self._compute_distance(
                X_array,
                centroids
            )

            labels = np.argmin(
                distances,
                axis=1
            )

            new_centroids = np.zeros_like(
                centroids
            )

            for cluster in range(
                self.n_clusters
            ):
                cluster_points = X_array[
                    labels == cluster
                ]

                if len(cluster_points) == 0:
                    new_centroids[cluster] = (
                        X_array[
                            rng.randint(
                                len(X_array)
                            )
                        ]
                    )
                else:
                    new_centroids[cluster] = (
                        np.median(
                            cluster_points,
                            axis=0
                        )
                    )

            shift = np.linalg.norm(
                new_centroids - centroids
            )

            centroids = new_centroids

            if shift < self.tol:
                break

        final_distances = self._compute_distance(
            X_array,
            centroids
        )

        labels = np.argmin(
            final_distances,
            axis=1
        )

        assigned_distances = (
            final_distances[
                np.arange(len(X_array)),
                labels
            ]
        )

        self.cluster_centers_ = centroids
        self.labels_ = labels
        self.inertia_ = float(
            np.sum(assigned_distances)
        )
        self.n_iter_ = iteration + 1

        return self
```

```python
k_values = list(range(1, 11))
manhattan_sae = []

for k in k_values:
    model_test = KMeansManhattan(
        n_clusters=k,
        max_iter=300,
        tol=1e-4,
        random_state=42
    )

    model_test.fit(
        X_scaled
    )

    manhattan_sae.append(
        model_test.inertia_
    )
```

```python
p1 = np.array([
    k_values[0],
    manhattan_sae[0]
])

p2 = np.array([
    k_values[-1],
    manhattan_sae[-1]
])

line_vec = p2 - p1
line_unit = (
    line_vec
    / np.linalg.norm(line_vec)
)

distances_elbow = []

for k, sae in zip(
    k_values,
    manhattan_sae
):
    point = np.array([
        k,
        sae
    ])

    vec = point - p1

    projection = (
        np.dot(
            vec,
            line_unit
        )
        * line_unit
    )

    perpendicular = (
        vec - projection
    )

    distances_elbow.append(
        np.linalg.norm(
            perpendicular
        )
    )

n_clusters = k_values[
    np.argmax(
        distances_elbow
    )
]

print(
    f'Jumlah Cluster Optimal: '
    f'k = {n_clusters}'
)
```

```python
plt.figure(
    figsize=(9, 5)
)

plt.plot(
    k_values,
    manhattan_sae,
    marker='o',
    markersize=8,
    linewidth=2.2,
    label='SAE Manhattan'
)

plt.axvline(
    x=n_clusters,
    linestyle='--',
    linewidth=2,
    label=f'Optimal k = {n_clusters}'
)

plt.scatter(
    [n_clusters],
    [
        manhattan_sae[
            n_clusters - 1
        ]
    ],
    s=180,
    zorder=5,
    edgecolor='black'
)

plt.title(
    'Elbow Method - K-Means Manhattan Distance',
    fontsize=13,
    fontweight='bold'
)

plt.xlabel(
    'Jumlah Cluster (K)'
)

plt.ylabel(
    'Sum of Absolute Errors (SAE)'
)

plt.xticks(
    k_values
)

plt.annotate(
    f'Elbow Point (k = {n_clusters})\n'
    f'SAE = '
    f'{manhattan_sae[n_clusters - 1]:,.2f}',
    xy=(
        n_clusters,
        manhattan_sae[
            n_clusters - 1
        ]
    ),
    xytext=(
        n_clusters + 0.8,
        manhattan_sae[
            n_clusters - 1
        ]
        + (
            max(manhattan_sae)
            - min(manhattan_sae)
        ) * 0.1
    ),
    arrowprops=dict(
        shrink=0.08,
        width=2,
        headwidth=7
    ),
    bbox=dict(
        boxstyle='round,pad=0.4'
    ),
    fontweight='bold'
)

plt.grid(
    True,
    linestyle=':',
    alpha=0.6
)

plt.legend()
plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        'elbow_manhattan.png'
    ),
    dpi=300
)

plt.show()
```

3.2 Training K-Means Manhattan


```python
tracemalloc.start()

start_time = time.perf_counter()

kmeans_manhattan = KMeansManhattan(
    n_clusters=n_clusters,
    max_iter=300,
    tol=1e-4,
    random_state=42
)

kmeans_manhattan.fit(
    X_scaled
)

runtime = (
    time.perf_counter()
    - start_time
)

current_memory, peak_memory = (
    tracemalloc.get_traced_memory()
)

tracemalloc.stop()

peak_memory_mb = (
    peak_memory / (1024 ** 2)
)

labels = (
    kmeans_manhattan.labels_
)

df['KMeans_Cluster'] = labels
```

```python
iterations = (
    kmeans_manhattan.n_iter_
)

sae_final = (
    kmeans_manhattan.inertia_
)

print(
    f'Runtime        : '
    f'{runtime:.5f} detik'
)

print(
    f'Iterations     : '
    f'{iterations}'
)

print(
    f'SAE Manhattan  : '
    f'{sae_final:.2f}'
)

print(
    f'Peak Memory    : '
    f'{peak_memory_mb:.2f} MB'
)
```
