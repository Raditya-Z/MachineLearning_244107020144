```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# **NO 1 - Pemisahan Variabel**

```python
# Menghapus kolom yang tidak digunakan
df = df.drop(columns=['id', 'Unnamed: 32', ])

df.head()
```

**Penjelasan No 1**

Pada dataset Wisconsin Breast Cancer, variabel dipisahkan berdasarkan kegunaannya:

`id` tidak digunakan karena hanya sebagai identitas data.

`Unnamed: 32` tidak digunakan karena merupakan kolom kosong.

`diagnosis` digunakan sebagai target untuk menentukan kanker Malignant (M) atau Benign (B).


> Kolom numerik lainnya digunakan sebagai fitur karena berisi karakteristik sel/tumor yang membantu proses diagnosis.



# **No. 2 - Encoding Kolom diagnosis**

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])
```

```python
df.head(100)
```

**Penjelasan
No. 2 - Encoding Kolom diagnosis**

Kolom diagnosis masih berupa data kategori berupa huruf, sehingga perlu diubah menjadi angka agar dapat diproses oleh machine learning.

Hasil encoding:

- B (Benign) → 0
- M (Malignant) → 1

Dengan encoding, data diagnosis menjadi berbentuk numerik sehingga lebih mudah digunakan dalam proses pemodelan.

# **NO 3 - Standardisasi kolom numerik.**

```python
from sklearn.preprocessing import StandardScaler

std = StandardScaler()

kolom_numerik = df.drop(columns=['diagnosis']).columns
df[kolom_numerik] = std.fit_transform(df[kolom_numerik])

df.head(100)
```

**Penjelasan No 3 - Standarisasi kolom numerik**

Kode tersebut melakukan standardisasi pada seluruh fitur numerik menggunakan` StandardScaler`. Kolom `diagnosis` tidak ikut distandardisasi karena merupakan variabel target. Standardisasi membuat fitur memiliki skala yang seragam sehingga lebih sesuai digunakan dalam proses pembelajaran mesin.
